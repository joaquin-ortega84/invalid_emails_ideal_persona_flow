```python
import pandas as pd
import numpy as np

clay_leads = pd.read_csv('/mnt/c/Users/jortega/code/invalid_emails_ideal_persona_flow/new_sfdc_data/validated_leads_list_clay.csv')
clay_contacts = pd.read_csv('/mnt/c/Users/jortega/code/invalid_emails_ideal_persona_flow/new_sfdc_data/validated_contacts_list_clay.csv')

clay_leads['lead or contact'] = "lead"
clay_contacts['lead or contact'] = "contact"

clay_people = pd.concat([clay_leads, clay_contacts], axis=0, ignore_index=True)

display(clay_people.shape)
# display(clay_people.columns)

clay_people["Related Record ID"] = np.where(
    clay_people["lead or contact"] == "lead",
    clay_people["Lead_id"],
    clay_people["Contact_id"],
)

clay_people = clay_people[['Related Record ID','Email', 'Status (Zerobounce)', 'Sub Status (Zerobounce)',
       'Invalid Email (Output)', 'Description (Output)', 'lead or contact']]

display(clay_people.columns)
```
